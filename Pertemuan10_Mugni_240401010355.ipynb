**Langkah 1: Muat dan Eksplorasi Data**

```python
import pandas as pd

df = pd.read_csv("Telco-Customer-Churn.csv")

print("Ukuran Dataset:", df.shape)
display(df.head())

# Info dataset
print("\nInformasi Dataset")
df.info()

# Missing value
print("\nMissing Value")
print(df.isnull().sum())

# Proporsi kelas churn
print("\nProporsi Kelas Churn")
print(df["Churn"].value_counts())
print("\nPersentase Kelas Churn")
print(df["Churn"].value_counts(normalize=True))
```

**Langkah 2: Preprocessing**

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Memisahkan fitur dan target
X = df.drop("Churn", axis=1)
y = df["Churn"]

# Encoding fitur kategorikal
X = pd.get_dummies(X, drop_first=True)

# Encoding target (No = 0, Yes = 1)
y = y.map({"No": 0, "Yes": 1})

# Membagi data menjadi data latih dan data uji
X_tr, X_te, y_tr, y_te = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("Ukuran data latih:", X_tr.shape)
print("Ukuran data uji:", X_te.shape)

print("\nProporsi kelas data latih:")
print(y_tr.value_counts(normalize=True))

print("\nProporsi kelas data uji:")
print(y_te.value_counts(normalize=True))
```

Langkah 3: Latih Model

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42
)

rf.fit(X_tr, y_tr)
```

**Langkah 4: Evaluasi**

```python
from sklearn.metrics import classification_report, roc_auc_score

# Prediksi kelas
y_pred = rf.predict(X_te)

# Probabilitas kelas positif (Churn = 1)
y_prob = rf.predict_proba(X_te)[:, 1]

# Classification Report
print("Classification Report:")
print(classification_report(y_te, y_pred))

# ROC-AUC Score
roc_auc = roc_auc_score(y_te, y_prob)
print("ROC-AUC Score:", roc_auc)
```

**Langkah 5: Prediksi Probabilitas dan Simpulkan**

```python
# Menghitung probabilitas churn untuk setiap pelanggan pada data uji
churn_probability = rf.predict_proba(X_te)[:, 1]

# Menampilkan 10 probabilitas pertama
print("10 Probabilitas Churn Pertama:")
print(churn_probability[:10])
```

**Kesimpulan:**

Hasil pengujian menunjukkan bahwa model Random Forest mampu mengklasifikasikan pelanggan yang berpotensi melakukan churn dengan cukup baik. Penggunaan `class_weight="balanced"` membantu model menangani perbedaan jumlah data pada setiap kelas. Nilai probabilitas yang dihasilkan dari `predict_proba()` dapat digunakan untuk mengetahui tingkat risiko churn setiap pelanggan. Dengan demikian, model ini dapat menjadi pendukung dalam menentukan strategi untuk mempertahankan pelanggan.
